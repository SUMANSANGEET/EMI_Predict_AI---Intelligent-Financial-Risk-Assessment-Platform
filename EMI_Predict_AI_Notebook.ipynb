
## 2.🔎 Data Quality Assessment & Cleaning 📊

Real-world exports are messy. A quick audit surfaces three concrete issues that need fixing before any modeling:

1. **Malformed numerics** — `age`, `monthly_salary`, and `bank_balance` were ingested as text and contain corrupted values like `"64300.0.0"` (double decimal points) instead of clean floats.
2. **Inconsistent categorical casing** — `gender` contains `Female / female / F / FEMALE / Male / male / M / MALE` instead of two clean labels.
3. **Missing values** — `education`, `monthly_rent`, `credit_score`, `bank_balance`, and `emergency_fund` all have ~0.5–0.6% nulls.

A raw duplicate-row check reports 0 duplicates — but this is misleading: some rows are true duplicates whose `age` / `monthly_salary` / `bank_balance` values were corrupted *differently* (e.g. `"58.0.0"` vs `"58.0"`), which hides them from a naive string-level check. Once those columns are cleaned to numeric form, the real duplicates surface — a classic case of corruption masking duplication.


````python
# --- Audit before cleaning ---
quality_report = pd.DataFrame({
    'missing_count': df.isna().sum(),
    'missing_pct': (df.isna().sum() / len(df) * 100).round(2),
    'dtype': df.dtypes.astype(str)
})
quality_report = quality_report[quality_report['missing_count'] > 0].sort_values('missing_pct', ascending=False)
print(f"Duplicate rows: {df.duplicated().sum()}")
quality_report
````

````python

def clean_malformed_numeric(x):
    # Fixes values like '64300.0.0' -> 64300.0 by collapsing repeated decimal points.
    if pd.isna(x):
        return np.nan
    s = str(x)
    parts = s.split('.')
    if len(parts) > 2:
        s = parts[0] + '.' + parts[1]
    try:
        return float(s)
    except ValueError:
        return np.nan

for col in ['age', 'monthly_salary', 'bank_balance']:
    n_bad = df[col].apply(lambda x: pd.notna(x) and '.' in str(x) and str(x).count('.') > 1).sum()
    df[col] = df[col].apply(clean_malformed_numeric)
    print(f"'{col}': fixed {n_bad:,} malformed values -> now dtype {df[col].dtype}")
````

# ---🔤 Standardize gender casing/aliases ---

````python
gender_map = {'female': 'Female', 'f': 'Female', 'male': 'Male', 'm': 'Male'}
df['gender'] = df['gender'].str.strip().str.lower().map(gender_map)
print(df['gender'].value_counts())
````

# ---🧹 Duplicate check, post numeric-cleaning (see note above: corruption was masking true duplicates) ---

````python
n_dupes = df.duplicated().sum()
print(f"Duplicate rows found after numeric cleaning: {n_dupes:,}")
if n_dupes > 0:
    df = df.drop_duplicates().reset_index(drop=True)
    print(f"Dropped {n_dupes:,} duplicate rows -> {df.shape[0]:,} rows remaining")
````

# ---🩹 Impute missing values ---
# 🧮 Numeric: median imputation (robust to outliers, preserves distribution shape)

````python
numeric_missing = ['monthly_rent', 'credit_score', 'bank_balance', 'emergency_fund']
for col in numeric_missing:
    median_val = df[col].median()
    df[col] = df[col].fillna(median_val)
    print(f"'{col}': imputed with median = {median_val:,.1f}")
````

# 🏷️ Categorical: mode imputation

````python
df['education'] = df['education'].fillna(df['education'].mode()[0])
print(f"'education': imputed with mode = '{df['education'].mode()[0]}'")

print(f"\nRemaining missing values: {df.isna().sum().sum()}")
````

# --- 📉✂️ Outlier check & capping (winsorize at 1st/99th percentile for skew-prone financial columns) ---

````python
skew_cols = ['monthly_salary', 'bank_balance', 'emergency_fund', 'current_emi_amount', 'requested_amount']
fig, axes = plt.subplots(1, len(skew_cols), figsize=(20, 4))
for ax, col in zip(axes, skew_cols):
    sns.boxplot(y=df[col], ax=ax, color='#66c2a5')
    ax.set_title(col, fontsize=10)
plt.suptitle('Distribution Check — Financial Columns (before capping)', y=1.03)
plt.tight_layout()
plt.show()

for col in skew_cols:
    lo, hi = df[col].quantile(0.01), df[col].quantile(0.99)
    n_capped = ((df[col] < lo) | (df[col] > hi)).sum()
    df[col] = df[col].clip(lo, hi)
    print(f"'{col}': capped {n_capped:,} extreme values to [{lo:,.0f}, {hi:,.0f}]")
````

# ---🔁 Final duplicate sweep (capping extreme values to the 1st/99th percentile can
#    📉 collapse near-identical template rows into exact duplicates) ---

````python
n_dupes_final = df.duplicated().sum()
if n_dupes_final > 0:
    df = df.drop_duplicates().reset_index(drop=True)
    print(f"Dropped {n_dupes_final:,} additional duplicate rows surfaced by outlier capping -> {df.shape[0]:,} rows remaining")
````

## 4.⚙️ Feature Engineering

We derive financial ratios and risk-scoring features that a human underwriter would intuitively use — these consistently drive the largest gains in both model performance and interpretability for credit risk models.


````python

fe = df.copy()

# --- Total monthly financial burden ---
expense_cols = ['monthly_rent', 'school_fees', 'college_fees', 'travel_expenses',
                 'groceries_utilities', 'other_monthly_expenses']
fe['total_monthly_expenses'] = fe[expense_cols].sum(axis=1)
fe['total_monthly_obligations'] = fe['total_monthly_expenses'] + fe['current_emi_amount']

# --- Core affordability ratios ---
fe['debt_to_income_ratio'] = fe['current_emi_amount'] / fe['monthly_salary']
fe['expense_to_income_ratio'] = fe['total_monthly_expenses'] / fe['monthly_salary']
fe['disposable_income'] = fe['monthly_salary'] - fe['total_monthly_obligations']
fe['affordability_ratio'] = (fe['disposable_income'] / fe['monthly_salary']).clip(-2, 1)

# --- Savings & safety-net features ---
fe['savings_to_income_ratio'] = fe['bank_balance'] / fe['monthly_salary']
fe['emergency_fund_coverage_months'] = (fe['emergency_fund'] /
                                         fe['total_monthly_obligations'].replace(0, np.nan)).fillna(0).clip(0, 24)

# --- Requested-loan affordability ---
fe['naive_new_emi'] = fe['requested_amount'] / fe['requested_tenure']
fe['new_emi_to_income_ratio'] = fe['naive_new_emi'] / fe['monthly_salary']
fe['requested_to_annual_income_ratio'] = fe['requested_amount'] / (fe['monthly_salary'] * 12)

# --- Household / demographic interaction features ---
fe['income_per_dependent'] = fe['monthly_salary'] / (fe['dependents'] + 1)
fe['employment_stability_score'] = fe['years_of_employment'] / fe['age']

# --- Composite risk score (0-100, higher = lower risk) blending credit history,
#     affordability and employment stability — mirrors a simplified underwriting scorecard ---
credit_component = (fe['credit_score'] - 300) / (850 - 300) * 100
dti_component = (1 - fe['debt_to_income_ratio'].clip(0, 1)) * 100
stability_component = fe['employment_stability_score'].clip(0, 1) * 100
fe['risk_score'] = (0.5 * credit_component + 0.3 * dti_component + 0.2 * stability_component).clip(0, 100)

new_features = ['total_monthly_expenses', 'total_monthly_obligations', 'debt_to_income_ratio',
                 'expense_to_income_ratio', 'disposable_income', 'affordability_ratio',
                 'savings_to_income_ratio', 'emergency_fund_coverage_months', 'naive_new_emi',
                 'new_emi_to_income_ratio', 'requested_to_annual_income_ratio',
                 'income_per_dependent', 'employment_stability_score', 'risk_score']
print(f"Engineered {len(new_features)} new features:")
print(new_features)
fe[new_features].describe().T.round(2)
````

````python

fig, axes = plt.subplots(1, 3, figsize=(18, 4.5))
sns.boxplot(data=fe, x='emi_eligibility', y='risk_score', order=['Eligible','High_Risk','Not_Eligible'],
            palette=COLOR_MAP, ax=axes[0])
axes[0].set_title('Composite Risk Score by Eligibility', fontweight='bold')

sns.boxplot(data=fe, x='emi_eligibility', y='debt_to_income_ratio', order=['Eligible','High_Risk','Not_Eligible'],
            palette=COLOR_MAP, ax=axes[1])
axes[1].set_title('Debt-to-Income Ratio by Eligibility', fontweight='bold')
axes[1].set_ylim(0, 1.5)

sns.boxplot(data=fe, x='emi_eligibility', y='affordability_ratio', order=['Eligible','High_Risk','Not_Eligible'],
            palette=COLOR_MAP, ax=axes[2])
axes[2].set_title('Affordability Ratio by Eligibility', fontweight='bold')

plt.tight_layout()
plt.show()
print("The engineered risk_score cleanly separates the three eligibility classes — a strong sign it will be a top feature for the classifiers.")
````


### 4.1 🔄m Categorical Encoding & Numerical Scaling

- **Nominal categoricals** (`gender`, `marital_status`, `education`, `employment_type`, `company_type`, `house_type`, `existing_loans`, `emi_scenario`) → **one-hot encoded**.
- **Numeric features** → **standard-scaled** (zero mean, unit variance) — required for Logistic/Linear Regression; harmless (and skipped internally) for tree-based models, so we build one shared preprocessing pipeline for consistency and reproducibility.
- Encoding/scaling is fit **only on the training split** (below) to prevent data leakage.


````python
target_cls = 'emi_eligibility'
target_reg = 'max_monthly_emi'

categorical_features = ['gender', 'marital_status', 'education', 'employment_type',
                         'company_type', 'house_type', 'existing_loans', 'emi_scenario']

drop_cols = [target_cls, target_reg, 'naive_new_emi']  # naive_new_emi is a near-duplicate of the regression target's driver, dropped to avoid leakage-like trivial correlation
feature_columns = [c for c in fe.columns if c not in drop_cols]
numeric_features = [c for c in feature_columns if c not in categorical_features]

print(f"Numeric features ({len(numeric_features)}):", numeric_features)
print(f"\nCategorical features ({len(categorical_features)}):", categorical_features)
````


## 5.✂️ Train / Test Split & Preprocessing Pipeline

An 80/20 stratified split (stratified on `emi_eligibility`) preserves the class distribution in both sets. We build a single reusable `ColumnTransformer` so every model — classification and regression alike — is trained on an identical, leakage-free feature space.


````python

X = fe[feature_columns]
y_cls = fe[target_cls]
y_reg = fe[target_reg]

X_train, X_test, y_cls_train, y_cls_test, y_reg_train, y_reg_test = train_test_split(
    X, y_cls, y_reg, test_size=0.2, random_state=RANDOM_STATE, stratify=y_cls
)

print(f"Train set: {X_train.shape[0]:,} rows   |   Test set: {X_test.shape[0]:,} rows")
print("\nTrain class balance:")
print((y_cls_train.value_counts(normalize=True) * 100).round(1))

preprocessor = ColumnTransformer(transformers=[
    ('num', StandardScaler(), numeric_features),
    ('cat', OneHotEncoder(handle_unknown='ignore', drop='first'), categorical_features)
])

# Fit once, reuse the same encoded matrices across every model below
X_train_enc = preprocessor.fit_transform(X_train)
X_test_enc = preprocessor.transform(X_test)

ohe_feature_names = preprocessor.named_transformers_['cat'].get_feature_names_out(categorical_features)
all_feature_names = numeric_features + list(ohe_feature_names)
print(f"\nFinal encoded feature space: {X_train_enc.shape[1]} columns")

le_target = LabelEncoder()
y_cls_train_enc = le_target.fit_transform(y_cls_train)
y_cls_test_enc = le_target.transform(y_cls_test)
print("Encoded classes:", dict(zip(le_target.classes_, le_target.transform(le_target.classes_))))
````


## 6.🤖 Classification Modeling — 💳 EMI Eligibility Prediction

Three models spanning the interpretability–performance spectrum, each tracked in **MLflow** (parameters, metrics, confusion matrix, and the fitted model artifact):

| Model | Role |
|---|---|
| Logistic Regression | Interpretable baseline |
| Random Forest | Ensemble + feature importance |
| XGBoost | High-performance gradient boosting |

`class_weight='balanced'` (and XGBoost's equivalent sample weighting) compensates for the ~77/18/4 class imbalance identified in EDA.


````python

mlflow.set_experiment("EMIPredict_Classification")

classification_models = {
    'Logistic Regression': LogisticRegression(max_iter=1000, class_weight='balanced', random_state=RANDOM_STATE),
    'Random Forest': RandomForestClassifier(n_estimators=150, max_depth=12, min_samples_leaf=5,
                                             class_weight='balanced', n_jobs=-1, random_state=RANDOM_STATE),
    'XGBoost': XGBClassifier(n_estimators=250, max_depth=7, learning_rate=0.1, tree_method='hist',
                              eval_metric='mlogloss', n_jobs=-1, random_state=RANDOM_STATE)
}

cls_results = {}
cls_fitted = {}

sample_weight_map = y_cls_train_enc.copy().astype(float)
from sklearn.utils.class_weight import compute_sample_weight
xgb_sample_weights = compute_sample_weight(class_weight='balanced', y=y_cls_train_enc)

for name, model in classification_models.items():
    with mlflow.start_run(run_name=name):
        if name == 'XGBoost':
            model.fit(X_train_enc, y_cls_train_enc, sample_weight=xgb_sample_weights)
        else:
            model.fit(X_train_enc, y_cls_train_enc)

        y_pred = model.predict(X_test_enc)
        y_proba = model.predict_proba(X_test_enc)

        acc = accuracy_score(y_cls_test_enc, y_pred)
        prec = precision_score(y_cls_test_enc, y_pred, average='weighted', zero_division=0)
        rec = recall_score(y_cls_test_enc, y_pred, average='weighted', zero_division=0)
        f1 = f1_score(y_cls_test_enc, y_pred, average='weighted', zero_division=0)
        roc_auc = roc_auc_score(y_cls_test_enc, y_proba, multi_class='ovr', average='weighted')

        mlflow.log_param('model_type', name)
        mlflow.log_params({k: v for k, v in model.get_params().items() if isinstance(v, (int, float, str, bool)) or v is None})
        mlflow.log_metrics({'accuracy': acc, 'precision': prec, 'recall': rec, 'f1_score': f1, 'roc_auc': roc_auc})

        if name == 'XGBoost':
            mlflow.xgboost.log_model(model, artifact_path='model')
        else:
            mlflow.sklearn.log_model(model, artifact_path='model')

        cls_results[name] = {'Accuracy': acc, 'Precision': prec, 'Recall': rec, 'F1-Score': f1, 'ROC-AUC': roc_auc}
        cls_fitted[name] = (model, y_pred, y_proba)
        print(f"{name:22s} | Acc: {acc:.4f}  Prec: {prec:.4f}  Rec: {rec:.4f}  F1: {f1:.4f}  ROC-AUC: {roc_auc:.4f}")

cls_results_df = pd.DataFrame(cls_results).T.round(4)
cls_results_df
````

````python

fig = go.Figure()
metrics_to_plot = ['Accuracy', 'Precision', 'Recall', 'F1-Score', 'ROC-AUC']
for i, model_name in enumerate(cls_results_df.index):
    fig.add_trace(go.Bar(name=model_name, x=metrics_to_plot, y=cls_results_df.loc[model_name, metrics_to_plot],
                          marker_color=PALETTE[i], text=cls_results_df.loc[model_name, metrics_to_plot].round(3),
                          textposition='outside'))
fig.update_layout(barmode='group', title='Classification Model Comparison',
                   yaxis_title='Score', yaxis_range=[0, 1.05], template='plotly_white', height=480)
fig.show()
````


## 7.📈 Regression Modeling — Max Monthly EMI Prediction

The same train/test split and encoded feature space are reused to predict the continuous target `max_monthly_emi` — the maximum installment amount an applicant can safely afford.


````python

mlflow.set_experiment("EMIPredict_Regression")

regression_models = {
    'Linear Regression': LinearRegression(),
    'Random Forest': RandomForestRegressor(n_estimators=150, max_depth=12, min_samples_leaf=10, max_features='sqrt',
                                            n_jobs=-1, random_state=RANDOM_STATE),
    'XGBoost': XGBRegressor(n_estimators=300, max_depth=7, learning_rate=0.08, tree_method='hist',
                             n_jobs=-1, random_state=RANDOM_STATE)
}

reg_results = {}
reg_fitted = {}

for name, model in regression_models.items():
    with mlflow.start_run(run_name=name):
        model.fit(X_train_enc, y_reg_train)
        y_pred = model.predict(X_test_enc)

        rmse = np.sqrt(mean_squared_error(y_reg_test, y_pred))
        mae = mean_absolute_error(y_reg_test, y_pred)
        r2 = r2_score(y_reg_test, y_pred)
        mape = np.mean(np.abs((y_reg_test - y_pred) / y_reg_test.replace(0, np.nan))).clip(0, 5) * 100

        mlflow.log_param('model_type', name)
        mlflow.log_params({k: v for k, v in model.get_params().items() if isinstance(v, (int, float, str, bool)) or v is None})
        mlflow.log_metrics({'rmse': rmse, 'mae': mae, 'r2_score': r2, 'mape': mape})

        if name == 'XGBoost':
            mlflow.xgboost.log_model(model, artifact_path='model')
        else:
            mlflow.sklearn.log_model(model, artifact_path='model')

        reg_results[name] = {'RMSE': rmse, 'MAE': mae, 'R²': r2, 'MAPE (%)': mape}
        reg_fitted[name] = (model, y_pred)
        print(f"{name:22s} | RMSE: {rmse:,.1f}  MAE: {mae:,.1f}  R²: {r2:.4f}  MAPE: {mape:.1f}%")

reg_results_df = pd.DataFrame(reg_results).T.round(3)
reg_results_df
````

````python

fig = make_subplots(rows=1, cols=2, subplot_titles=('RMSE / MAE (₹, lower is better)', 'R² Score (higher is better)'))

for i, model_name in enumerate(reg_results_df.index):
    fig.add_trace(go.Bar(name=model_name, x=['RMSE', 'MAE'], y=reg_results_df.loc[model_name, ['RMSE', 'MAE']],
                          marker_color=PALETTE[i], showlegend=True), row=1, col=1)
    fig.add_trace(go.Bar(name=model_name, x=['R²'], y=[reg_results_df.loc[model_name, 'R²']],
                          marker_color=PALETTE[i], showlegend=False), row=1, col=2)

fig.update_layout(barmode='group', title='Regression Model Comparison', template='plotly_white', height=480)
fig.show()
````

````python

best_reg_name = reg_results_df['R²'].idxmax()
best_reg_model, best_reg_pred = reg_fitted[best_reg_name]
print(f"🏆 Best regression model (by R²): {best_reg_name}")

fig, axes = plt.subplots(1, 2, figsize=(15, 5.5))

sample_idx = np.random.choice(len(y_reg_test), 5000, replace=False)
axes[0].scatter(y_reg_test.values[sample_idx], best_reg_pred[sample_idx], alpha=0.25, s=12, color='#8da0cb')
lims = [0, max(y_reg_test.max(), best_reg_pred.max())]
axes[0].plot(lims, lims, 'r--', linewidth=1.5, label='Perfect Prediction')
axes[0].set_title(f'Actual vs Predicted — {best_reg_name}', fontweight='bold')
axes[0].set_xlabel('Actual Max Monthly EMI (₹)'); axes[0].set_ylabel('Predicted Max Monthly EMI (₹)')
axes[0].legend()

residuals = y_reg_test.values[sample_idx] - best_reg_pred[sample_idx]
axes[1].scatter(best_reg_pred[sample_idx], residuals, alpha=0.25, s=12, color='#fc8d62')
axes[1].axhline(0, color='r', linestyle='--', linewidth=1.5)
axes[1].set_title(f'Residual Plot — {best_reg_name}', fontweight='bold')
axes[1].set_xlabel('Predicted Max Monthly EMI (₹)'); axes[1].set_ylabel('Residual (Actual - Predicted)')

plt.tight_layout()
plt.show()
````


## 8.🏆 Feature Importance & Final Model Selection

````python

def get_importances(model, feature_names, top_n=15):
    if hasattr(model, 'feature_importances_'):
        imp = pd.Series(model.feature_importances_, index=feature_names)
    elif hasattr(model, 'coef_'):
        coef = model.coef_
        coef = np.abs(coef).mean(axis=0) if coef.ndim > 1 else np.abs(coef)
        imp = pd.Series(coef, index=feature_names)
    else:
        return None
    return imp.sort_values(ascending=False).head(top_n)

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

cls_imp = get_importances(best_cls_model, all_feature_names)
axes[0].barh(cls_imp.index[::-1], cls_imp.values[::-1], color='#66c2a5')
axes[0].set_title(f'Top 15 Features — {best_cls_name} (Eligibility)', fontweight='bold')

reg_imp = get_importances(best_reg_model, all_feature_names)
axes[1].barh(reg_imp.index[::-1], reg_imp.values[::-1], color='#fc8d62')
axes[1].set_title(f'Top 15 Features — {best_reg_name} (Max EMI)', fontweight='bold')

plt.tight_layout()
plt.show()
````
